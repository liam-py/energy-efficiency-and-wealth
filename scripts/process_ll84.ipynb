## 1. Import and inspect data

```python
import pandas as pd

# create a dataframe
df = pd.read_json("../data/raw/ll84_raw.json", orient="records")

print("A look at the raw data...")
df.head()
df.info()
df.describe()
```

## 2. Rename columns

```python
# standardize column names
df.columns = df.columns.str.strip().str.lower()

df = df.rename(columns={
    "nyc_borough_block_and_lot": "bbl",
    "site_eui_kbtu_ft": "eui"
})

print("Inspect row names")
df.head()
df.info()
```

## 3. Set Data Types

```python
# convert eui from str to numerical
df["eui"] = pd.to_numeric(df["eui"], errors="coerce")

print("Inspect new typing: ")
print(df.dtypes)
```

## 4. Standardize data entries

```python
print("test for null and duplicate rows pre-cleaning: ")
print("Null counts: ")
print(df.isnull().sum())
print(f'Starting duplicate rows: {df.duplicated().sum()}')
print(f"Starting N/A bbl rows: {(df['bbl'] == 'Not Available').sum()}")

# drop duplicate rows
df.drop_duplicates(inplace=True)

# drop entries with critical missing values
df.dropna(subset=['eui'], inplace=True) # drop if there is no eui
df = df[df['bbl'] != "Not Available"] # drop any row with out bbl

print("test for null and duplicate rows post-cleaning: ")
print("Null counts: ")
print(df.isnull().sum())
print(f'Remaining duplicate rows: {df.duplicated().sum()}')
print(f"Remaining N/A bbl rows: {(df['bbl'] == 'Not Available').sum()}")

```

## Final check

```python
print(df.shape)
print(df.isnull().sum())
print(df.dtypes)
df.head(10)
```

## Export final data

```python
df.to_csv('../data/clean/ll84_cleaned.csv', index=False)
print(f'Exported {len(df)} rows')
```
