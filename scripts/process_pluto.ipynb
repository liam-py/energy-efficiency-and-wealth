## Import and inspect data

```python
import pandas as pd

raw_df = pd.read_json("../data/raw/pluto_raw.json", orient="records", dtype=str)
df = raw_df.copy() # create a copy of the dataframe we can work with more freely

print("A look at the raw data...")
print()
df.info()
print()
df.describe()
```

## Rename & drop unnecessary columns

```python
df = df.rename(columns={'bct2020': 'bct'}) # borough census tract
df.columns = df.columns.str.strip().str.lower() # standardize column names

print("Inspect row names")
print()
df.info()
```

## 3. Standardize data entries

```python
print("test for null and duplicate rows pre-cleaning: ")
print("Null counts: ")
print(df.isnull().sum())
print(f"Starting duplicate rows: {df.duplicated().sum()}")

# format to simple 10 digit bbl to match LL84 data
df['bbl'] = df['bbl'].str.split('.').str[0] # since not a numeric, just split string at dec pt and take first part

print("Simple 10 digit bbl on first 3 entries: ")
df.head(3)
```

## 4. Final check

```python
print("data shape: ")
print(df.shape)
print()

print("null counts:")
print(df.isnull().sum())
print()

print("data types:")
print(df.dtypes)
print()

print("first few entries:")
df.head(10)
```

## 5. Export data

```python
df.to_pickle('../data/clean/pluto_cleaned.pkl')
print(f'Exported {len(df)} rows as pkl')
df.to_csv('../data/clean/pluto_cleaned.csv')
print(f'Exported {len(df)} rows as csv')
```
