## Import and inspect data

```python
import pandas as pd

raw_df = pd.read_json("../data/raw/pluto_raw.json", orient="records", dtype=str)
df = raw_df.copy() # create a copy of the dataframe we can work with more freely

print("A look at the raw data...")
print()
df.info()
print()
df.describe()
```

## Rename & drop unnecessary columns

```python
df.rename(columns={'bct2020': 'bct'}) # borough census tract
df.columns = df.columns.str.strip().str.lower() # standardize column names

print("Inspect row names")
print()
df.info()
```
